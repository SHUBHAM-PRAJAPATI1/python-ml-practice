## Loan EMI Calculator
Calculates Equated Monthly Installment (EMI), total payment, and total interest for a loan.

```python
def calculate_emi(principal, annual_rate, tenure_years):
    """
    Calculate EMI, total payment, and total interest.
    
    principal: loan amount (P)
    annual_rate: annual interest rate in % (e.g. 8.5)
    tenure_years: loan duration in years
    """
    monthly_rate = annual_rate / (12 * 100)
    tenure_months = tenure_years * 12

    if monthly_rate == 0:
        emi = principal / tenure_months
    else:
        emi = (principal * monthly_rate * (1 + monthly_rate) ** tenure_months) / \
              ((1 + monthly_rate) ** tenure_months - 1)

    total_payment = emi * tenure_months
    total_interest = total_payment - principal

    return round(emi, 2), round(total_payment, 2), round(total_interest, 2)
```

```python
test_cases = [
    (500000, 8.5, 5),    # 5L loan, 8.5% rate, 5 years
    (1000000, 7.2, 10),  # 10L loan, 7.2% rate, 10 years
    (200000, 0, 2),      # edge case: 0% interest
]

for principal, rate, years in test_cases:
    emi, total, interest = calculate_emi(principal, rate, years)
    print(f"Principal: ₹{principal:,} | Rate: {rate}% | Tenure: {years}yr")
    print(f"  EMI: ₹{emi:,} | Total Payment: ₹{total:,} | Total Interest: ₹{interest:,}")
    print()
```

```python
def get_user_loan_details():
    principal = float(input("Enter loan amount: "))
    annual_rate = float(input("Enter annual interest rate (%): "))
    tenure_years = int(input("Enter tenure (years): "))
    
    emi, total, interest = calculate_emi(principal, annual_rate, tenure_years)
    print(f"\nMonthly EMI: ₹{emi:,}")
    print(f"Total Payment: ₹{total:,}")
    print(f"Total Interest Payable: ₹{interest:,}")

# get_user_loan_details()  # uncomment to run interactively
```
